```python
import tellurium as te
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Define the model
rr = te.loada('''
    E_h' = e1*(E_h + E_d)*(1 - (E_h + E_d)/K) - e2*E_h - e3*E_h + e4*E_d;
    E_d' = -e2*E_d + e3*E_h - e4*E_d;

    E_h = 0.2; E_d = 0;

    e1 = 0.0315;
    e2 = 0.001;
    e3 = 0.005;
    e4 = 0.002;
    K = 1.0;
''')

# Setup for plotting
plt.figure(figsize=(7, 6))
start_vals = np.linspace(0.0, 1, 4)  # fewer initial conditions

# Trajectories
for Eh0 in start_vals:
    for Ed0 in start_vals:
        rr.reset()
        rr.E_h = Eh0
        rr.E_d = Ed0
        result = rr.simulate(0, 3000, 1000, ["E_h", "E_d"])
        plt.plot(result[:, 0], result[:, 1], 'b', alpha=0.7)

# Mesh grid for nullclines
Eh_vals = np.linspace(0.01, 1.0, 200)
Ed_vals = np.linspace(0.01, 1.0, 200)
Eh_mesh, Ed_mesh = np.meshgrid(Eh_vals, Ed_vals)

# Parameters
e1 = 0.0315; e2 = 0.001; e3 = 0.005; e4 = 0.002; K = 1.0
Etot = Eh_mesh + Ed_mesh
dEh_dt = e1 * Etot * (1 - Etot / K) - e2 * Eh_mesh - e3 * Eh_mesh + e4 * Ed_mesh
dEd_dt = -e2 * Ed_mesh + e3 * Eh_mesh - e4 * Ed_mesh

# Nullclines
plt.contour(Eh_mesh, Ed_mesh, dEh_dt, levels=[0], colors='black', linestyles='--', linewidths=2)
plt.contour(Eh_mesh, Ed_mesh, dEd_dt, levels=[0], colors='black', linestyles=':', linewidths=2)

# Custom legend
custom_lines = [
    Line2D([0], [0], color='black', linestyle='--', linewidth=2, label=r"$\frac{dE_h}{dt} = 0$"),
    Line2D([0], [0], color='black', linestyle=':', linewidth=2, label=r"$\frac{dE_d}{dt} = 0$")
]
plt.legend(handles=custom_lines, loc='best', fontsize=12)

# Labels and layout
plt.xlabel(r"$E_h$", fontsize=14)
plt.ylabel(r"$E_d$", fontsize=14)
#plt.title("Phase Plane with Nullclines", fontsize=16)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.grid(True)
plt.tight_layout()
plt.savefig("phase_plane_nullclines_bw.pdf")
plt.show()
```

```python
import tellurium as te
import rrplugins
import matplotlib.pyplot as plt

# Define model
r = te.loada('''
    E_h' = e1*(E_h + E_d)*(1 - (E_h + E_d)/K) - e2*E_h - e3*E_h + e4*E_d;
    E_d' = -e2*E_d + e3*E_h - e4*E_d;

    E_h = 0.2; E_d = 0;

    e1 = 0.0315;
    e2 = 0.001;
    e3 = 0.005;
    e4 = 0.002;
    K = 1.0;
''')
```

```python
auto = rrplugins.Plugin("tel_auto2000")

# Set the SBML model
auto.setProperty("SBML", r.getCurrentSBML())

# Set the continuation parameter (e.g., e1 or e3 or another parameter)
auto.setProperty("PrincipalContinuationParameter", "e1")

# Set scan limits
auto.setProperty("RL0", 0.0)   # Lower limit
auto.setProperty("RL1", 0.1)   # Upper limit

# Set optional pre-simulation to find a good steady state
auto.setProperty("PreSimulation", "True")
auto.setProperty("PreSimulationDuration", 10)

# Set scan direction
auto.setProperty("ScanDirection", "Positive")

# Optional: tighter tolerance
auto.setProperty("Tolerance", 1e-6)
```
